```python


import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy import stats

# Dataset
scores = [78, 85, 92, 65, 72, 88, 90, 78, 85, 80]

# Convert to NumPy array
data = np.array(scores)

# Sort the data
sorted_data = np.sort(data)

print("Original Data:", scores)
print("Sorted Data:", sorted_data)


mean = np.mean(data)
median = np.median(data)

# Find mode
mode_result = stats.mode(data, keepdims=False)
mode = mode_result.mode

print("\n--- A. Mean, Median, Mode ---")
print("Mean   =", mean)
print("Median =", median)
print("Mode   =", mode)

# --------------------------------------------
# 2. RANGE, VARIANCE, STANDARD DEVIATION, IQR
# --------------------------------------------

# Range
data_range = np.max(data) - np.min(data)

# Population variance and standard deviation
variance = np.var(data)
std_dev = np.std(data)

# Quartiles
Q1 = np.percentile(data, 25)
Q3 = np.percentile(data, 75)

# Interquartile Range
IQR = Q3 - Q1

print("\n--- B. Range, Variance, Standard Deviation, IQR ---")
print("Range =", data_range)
print("Population Variance =", variance)
print("Population Standard Deviation =", std_dev)
print("Q1 =", Q1)
print("Q3 =", Q3)
print("IQR =", IQR)



skewness = stats.skew(data, bias=True)

# Excess kurtosis
kurtosis = stats.kurtosis(data, fisher=True, bias=True)

print("\n--- C. Skewness and Kurtosis ---")
print("Skewness =", skewness)
print("Excess Kurtosis =", kurtosis)



minimum = np.min(data)
maximum = np.max(data)

print("\n--- D. Five-Point Summary ---")
print("Minimum =", minimum)
print("Q1      =", Q1)
print("Median  =", median)
print("Q3      =", Q3)
print("Maximum =", maximum)



lower_fence = Q1 - 1.5 * IQR
upper_fence = Q3 + 1.5 * IQR

outliers = data[(data < lower_fence) | (data > upper_fence)]

print("\n--- Outlier Detection ---")
print("Lower Fence =", lower_fence)
print("Upper Fence =", upper_fence)

if len(outliers) == 0:
    print("Outliers = None")
else:
    print("Outliers =", outliers)



plt.figure(figsize=(8, 5))

plt.boxplot(data)

plt.title("Box Plot of Student Scores")
plt.ylabel("Scores")
plt.grid(axis='y', linestyle='--', alpha=0.5)

plt.show()
```
